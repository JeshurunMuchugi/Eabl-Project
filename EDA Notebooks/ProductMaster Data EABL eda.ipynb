## Find duplicates in categorical vairables
### i) Brand-family redundancy check (OLD_BRAND, EB_Brand_SIT,
###    MS_Brand_SIT, TM_Brand vs Brand)
###    Logic: if a field is a strict "rollup" of Brand, then every
###    Brand value maps to exactly ONE value of that field.
###    If nunique() > 1 per Brand group, it's carrying independent info
###    and dropping it loses something.


```python

brand_fields = ['OLD_BRAND', 'EB_Brand_SIT', 'MS_Brand_SIT', 'TM_Brand']
for col in brand_fields:
    mapping_consistency = products.groupby('Brand')[col].nunique()
    clean_rollup_pct = (mapping_consistency == 1).mean()
    print(f"{col}: {clean_rollup_pct:.2%} of Brands map to exactly one {col} value "
          f"(100% = safe to drop, lower = it's adding real info)")
# Also check the reverse direction for TM_Brand specifically,
# since TM_Brand (82 values) is coarser than Brand (138 values) —
# confirm it's a clean many-to-one rollup, not overlapping categories
tm_check = products.groupby('TM_Brand')['Brand'].nunique().sort_values(ascending=False)
print(tm_check.head(10))  # if TM_Brand groups contain many distinct Brands, that's expected (coarser)
```

```python
# ---------------------------------------------------------
# 3. Category taxonomy overlap (Category2, Sub_Category1,
#    Sub_Category2, New_Sub_Category2, CATMAN_Category,
#    Battleground_Category vs Category/Sub_Category)
# ---------------------------------------------------------
cat_fields = ['Category2', 'Sub_Category1', 'Sub_Category2',
              'New_Sub_Category2', 'CATMAN_Category', 'Battleground_Category']

for col in cat_fields:
    mapping_consistency = products.groupby('Category')[col].nunique()
    clean_rollup_pct = (mapping_consistency == 1).mean()
    print(f"{col}: {clean_rollup_pct:.2%} of Categories map to exactly one {col} value")
    print(f"  -> {col} has {products[col].nunique()} unique values vs "
          f"Category's {products['Category'].nunique()} / Sub_Category's {products['Sub_Category'].nunique()}")

# Direct visual check of how each field slices Category
for col in cat_fields:
    print(f"\n--- Category x {col} ---")
    print(pandas.crosstab(products['Category'], products[col]))
```

```python
col_a, col_b = "Brand", "OLD_BRAND"

# ---------------------------------------------------------
# 1. Pairwise comparison summary
# ---------------------------------------------------------
def compare_columns(df, col_a, col_b):
    equal_mask = df[col_a].astype(str).str.strip().str.lower() == \
                 df[col_b].astype(str).str.strip().str.lower()
    pct_equal = equal_mask.mean() * 100

    map_a_to_b = df.groupby(col_a)[col_b].nunique()
    map_b_to_a = df.groupby(col_b)[col_a].nunique()
    a_determines_b = (map_a_to_b <= 1).all()
    b_determines_a = (map_b_to_a <= 1).all()

    if a_determines_b and b_determines_a:
        relationship = "1:1 (fully redundant)"
    elif a_determines_b:
        relationship = f"{col_a} -> {col_b} many-to-1 ({col_a} is finer-grained)"
    elif b_determines_a:
        relationship = f"{col_b} -> {col_a} many-to-1 ({col_b} is finer-grained)"
    else:
        relationship = "no consistent mapping (independent info)"

    return {
        "col_a": col_a, "col_b": col_b,
        "pct_exact_match": round(pct_equal, 1),
        "n_unique_a": df[col_a].nunique(dropna=True),
        "n_unique_b": df[col_b].nunique(dropna=True),
        "relationship": relationship,
    }

result = compare_columns(products, col_a, col_b)
print(pd.DataFrame([result]).to_string(index=False))

# ---------------------------------------------------------
# 2. Crosstab: for each OLD_BRAND, which Brand values show up?
# ---------------------------------------------------------
crosstab = pd.crosstab(products[col_b], products[col_a])
print("\nCrosstab (rows = OLD_BRAND, columns = Brand counts):")
print(crosstab)

# ---------------------------------------------------------
# 3. Row-level agreement / disagreement
# ---------------------------------------------------------
agree_mask = products[col_a].astype(str).str.strip().str.lower() == \
             products[col_b].astype(str).str.strip().str.lower()

print(f"\nRows where values match exactly: {agree_mask.sum()}/{len(products)} "
      f"({agree_mask.mean()*100:.1f}%)")

mismatches = products.loc[~agree_mask, ["Product_Code", "Product_Description", col_b, col_a]]
print(f"\nMismatched rows ({len(mismatches)}):")
print(mismatches.sort_values(col_b).to_string(index=False))

# ---------------------------------------------------------
# 4. Distinct old->new rename pairs (collapsed, easier to read)
# ---------------------------------------------------------
rename_pairs = mismatches[[col_b, col_a]].drop_duplicates().sort_values(col_b)
print(f"\nDistinct OLD_BRAND -> Brand pairs ({len(rename_pairs)}):")
print(rename_pairs.to_string(index=False))

```

```python
# ===========================================================
# 1. Product_Code uniqueness — BLOCKING, run this before anything else
# ===========================================================
code_dupe_mask = products['Product_Code'].duplicated(keep=False)
n_dupes = code_dupe_mask.sum()
print(f"{n_dupes} rows share a Product_Code with another row")

if n_dupes > 0:
    dupe_detail = products.loc[
        code_dupe_mask,
        ['Product_Code', 'Product_Description', 'Status'] +
        [c for c in products.columns if 'EX_KBL_Price_Per_Case' in c]
    ].sort_values('Product_Code')
    print(dupe_detail.to_string(index=False))


    def resolve_duplicate_codes(df):
        resolved_rows = []
        flagged = []
        for code, grp in df.groupby('Product_Code'):
            if len(grp) == 1:
                resolved_rows.append(grp.iloc[0])
                continue
            active = grp[grp['Status'] == 'Active']
            if len(active) == 1:
                resolved_rows.append(active.iloc[0])
            else:
                candidates = active if len(active) > 1 else grp
                chosen = candidates.sort_values('FY26_EX_KBL_Price_Per_Case', ascending=False).iloc[0]
                resolved_rows.append(chosen)
            if len(grp) > 1:
                flagged.append(grp)
        resolved = pandas.DataFrame(resolved_rows).reset_index(drop=True)
        flagged_df = pandas.concat(flagged) if flagged else df.iloc[0:0].copy()
        return resolved, flagged_df

    products, code_dupes_flagged = resolve_duplicate_codes(products)
    print(f"\nResolved -> {len(products)} rows, Product_Code now unique: "
          f"{products['Product_Code'].is_unique}")
    print(f"\nRows that needed a tiebreak decision (verify these manually):")
    print(code_dupes_flagged[['Product_Code', 'Product_Description', 'Status']].to_string(index=False))
else:
    print("Product_Code is already unique — safe to use as the join key.")

assert products['Product_Code'].is_unique, "Product_Code still has duplicates — do not merge yet."
```
